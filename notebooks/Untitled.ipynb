```python
from pathlib import Path
import pandas as pd

RAW_DIR = Path("/home/RYVEN/tweets/data/raw/twcs")

def inspect_csv(path: Path) -> None:
    print(f"\n{'=' * 80}")
    print(f"FILE: {path}")
    print(f"SIZE: {path.stat().st_size / 1e6:.2f} MB")

    try:
        df = pd.read_csv(path, nrows=5)

        print("\nColumns:")
        for column in df.columns:
            print(f"  - {column}")

        print("\nSample:")
        print(df.to_string(index=False))

    except Exception as exc:
        print(f"Could not read as CSV: {exc}")


def main() -> None:
    files = sorted(RAW_DIR.iterdir())

    if not files:
        print(f"No files found in {RAW_DIR}")
        return

    print("Files found:")
    for path in files:
        print(f"  - {path}")

    for path in files:
        if path.suffix.lower() == ".csv":
            inspect_csv(path)


if __name__ == "__main__":
    main()
```
